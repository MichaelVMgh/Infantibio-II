```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np

# We load the data
df = pd.read_csv("merged_abundance_table_species_transposed.txt", sep="\t")

# Filter by the feeding group
df = df[df["HistoryAtTimePoint"] == "D"]

# Define the order of timepoints
timepoint_order = ["1mo", "3mo", "7mo", "12mo"]
df['TimePoint'] = pd.Categorical(df['TimePoint'], categories=timepoint_order, ordered=True)

# Sort by TimePoint and then by ID
df = df.sort_values(by=["TimePoint", "ID"])

# Define taxa
taxa = ["s__Bifidobacterium_bifidum", "s__Bifidobacterium_breve", "t__subsp.infantis", "t__subsp.longum"]

# Italic legend labels
taxa_labels = {
    "s__Bifidobacterium_bifidum": r"$\it{B. bifidum}$",
    "s__Bifidobacterium_breve": r"$\it{B. breve}$",
    "t__subsp.infantis": r"$\it{BL}$ subsp. $\it{infantis}$",
    "t__subsp.longum": r"$\it{BL}$ subsp. $\it{longum}$",
}

colors = sns.color_palette("deep", n_colors=len(taxa))

palette = {
    "t__subsp.infantis": colors[0],      # blue
    "t__subsp.longum": colors[1],        # orange
    "s__Bifidobacterium_breve": colors[2], # green
    "s__Bifidobacterium_bifidum": colors[3]  # red
}

sns.set_style("white")

# Melt the DataFrame
df_melted = df.melt(id_vars=["TimePoint"], value_vars=taxa, var_name="Taxa", value_name="Relative Abundance")

# Set up the figure and axes
fig, ax = plt.subplots(figsize=(15, 10))

# We generate the boxplot
sns.boxplot(
    data=df_melted,
    x="TimePoint",
    y="Relative Abundance",
    hue="Taxa",
    palette=palette,
    linewidth=2,
    hue_order=taxa,           
    showfliers=False
)

# Generate the stripplot
sns.stripplot(
    data=df_melted,
    x="TimePoint",
    y="Relative Abundance",
    hue="Taxa",
    palette=palette,
    dodge=True,
    jitter=True,
    alpha=0.6,
    marker="o",
    size=8,
    hue_order=taxa            
)

# Labels and title
plt.xlabel("Time point (months)", fontsize=36)
plt.ylabel("Relative Abundance", fontsize=36)
plt.title("Formula D", fontsize=46)

# X-axis labels (remove "mo")
new_labels = [label.replace("mo", "") for label in timepoint_order]
plt.xticks(ticks=range(len(timepoint_order)), labels=new_labels, fontsize=30)
plt.yticks(fontsize=30)
plt.ylim(0, 100)

# Legend with italic formatting
handles, labels = ax.get_legend_handles_labels()
legend_labels = [taxa_labels[label] for label in labels[:4]]
ax.legend(handles[:4], legend_labels, fontsize=26, title_fontsize=28)

# Remove top and right spines
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
#fig.savefig("D_bifido_boxplots_27.05.2026.png", dpi=800, bbox_inches='tight')
plt.show()
```
