```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

#We load the data
df = pd.read_csv("merged_abundance_table_genus_transposed.txt", sep="\t")

#We do not consider the "mother" and "father" samples (only infant samples)
df = df[~df["TimePoint"].isin(["mother", "father"])]

#Select and rename target genera
selected_genera = {
    "g__Bifidobacterium": "Bifidobacterium",
    "g__Veillonella": "Veillonella",
    "g__Escherichia": "Escherichia",
    "g__Bacteroides": "Bacteroides",
    "g__Phocaeicola": "Phocaeicola"
}

#We keep only the relevant columns
df = df[['HistoryAtTimePoint'] + list(selected_genera.keys())]
df = df.rename(columns=selected_genera)

#Group by feeding history and calculate median abundance
df_median = df.groupby("HistoryAtTimePoint").median()

# Sort feeding groups
order = ["A", "B", "C", "D", "BF", "feeding_at_24mo"]
df_median = df_median.loc[order]

# We clean the x-axis labels
df_median.index = df_median.index.str.replace("feeding_at_24mo", "infant_24mo")

#We define colors for each genus
custom_colors = {
    "Bifidobacterium": "#8da0cb",   # blueish
    "Escherichia": "#f8993a",       # orange
    "Veillonella": "#abc69d",       # greenish
    "Bacteroides": "#771155",       # purple
    "Phocaeicola": "#d37947"        # brownish
}

# Match colors to column order
colors = [custom_colors.get(col, "#cccccc") for col in df_median.columns]

# We generate the plot
plt.figure(figsize=(16, 8))
ax = df_median.plot(
    kind="bar",
    stacked=True,
    color=colors,
    figsize=(16, 8),
    edgecolor='black'
)

# labels and title
plt.ylabel("Median relative abundance (%)", fontsize=28)
plt.xlabel("Feeding group", fontsize=28)
plt.title("Top Genera by Feeding", fontsize=32)
plt.xticks(rotation=0, fontsize=22)
plt.yticks(fontsize=22)
plt.ylim(0, 100)

# We genera names in italic
genus_labels = {
    "Bifidobacterium": r"$\it{Bifidobacterium}$",
    "Veillonella": r"$\it{Veillonella}$",
    "Escherichia": r"$\it{Escherichia}$",
    "Bacteroides": r"$\it{Bacteroides}$",
    "Phocaeicola": r"$\it{Phocaeicola}$"
}

handles, labels = ax.get_legend_handles_labels()
italic_labels = [genus_labels.get(label, label) for label in labels]

ax.legend(
    handles,
    italic_labels,
    title="Genus",
    title_fontsize=24,
    fontsize=22,
    frameon=True,
    edgecolor='lightgrey',
    loc="upper right",
    bbox_to_anchor=(1.05, 1.1)
)

# We remove top and right spines
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Save
#plt.savefig("genus_stacked_median_abundances_histp_30.05.2026.png", dpi=800, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# We load the data
df = pd.read_csv("merged_abundance_table_genus_transposed.txt", sep="\t")

# We do not consider the "mother" and "father" samples (only infant samples)
df = df[~df["TimePoint"].isin(["mother", "father"])]

# Select and rename target genera
selected_genera = {
    "g__Bifidobacterium": "Bifidobacterium",
    "g__Veillonella": "Veillonella",
    "g__Escherichia": "Escherichia",
    "g__Bacteroides": "Bacteroides",
    "g__Phocaeicola": "Phocaeicola"
}

df = df[['TimePoint'] + list(selected_genera.keys())]
df = df.rename(columns=selected_genera)

# We group by TimePoint and calculate median abundance
df_median = df.groupby("TimePoint").median()

# Sort TimePoint in the desired order
order = ["1mo", "3mo", "7mo", "12mo", "24mo"]
df_median = df_median.loc[order]

# Simplify x-axis labels (e.g., "1mo" → "1")
df_median.index = df_median.index.str.replace(r"(\d+)mo", r"\1", regex=True)

# We define custom colors for each genus
custom_colors = {
    "Bifidobacterium": "#8da0cb",   # blueish
    "Escherichia": "#f8993a",       # orange
    "Veillonella": "#abc69d",       # greenish
    "Bacteroides": "#771155",       # purple
    "Phocaeicola": "#d37947"        # brownish
}

# Match colors to the column order
colors = [custom_colors.get(col, "#cccccc") for col in df_median.columns]

# We generate the plot
plt.figure(figsize=(16, 10))
ax = df_median.plot(
    kind="bar",
    stacked=True,
    color=colors,
    figsize=(16, 8),
    edgecolor='black',
    width=0.6
)

# Customize labels and title
plt.ylabel("Median relative abundance (%)", fontsize=28)
plt.xlabel("Time point (months)", fontsize=28)
plt.title("Top Genera over time", fontsize=32)
plt.xticks(rotation=0, fontsize=22)
plt.yticks(fontsize=22)
plt.ylim(0, 100)

# Genus names in italic
genus_labels = {
    "Bifidobacterium": r"$\it{Bifidobacterium}$",
    "Veillonella": r"$\it{Veillonella}$",
    "Escherichia": r"$\it{Escherichia}$",
    "Bacteroides": r"$\it{Bacteroides}$",
    "Phocaeicola": r"$\it{Phocaeicola}$"
}

handles, labels = ax.get_legend_handles_labels()
italic_labels = [genus_labels.get(label, label) for label in labels]

ax.legend(
    handles,
    italic_labels,
    title="Genus",
    title_fontsize=20,
    fontsize=18,
    frameon=True,
    edgecolor='lightgrey'
)

# Clean plot figure
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# Save the plot
#plt.savefig("genus_stacked_median_abundances_27.05.2026.png", dpi=800)

# Display the plot
plt.show()
```
